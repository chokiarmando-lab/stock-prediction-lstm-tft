# Pengambilan Data Saham LQ45

```python
stocks = {
    "BBCA.JK": "Bank Central Asia",
    "TLKM.JK": "Telkom Indonesia",
    "ASII.JK": "Astra International",
    "UNVR.JK": "Unilever Indonesia",
    "ICBP.JK": "Indofood CBP"
}
```

# Proses Preprocessing Data

```python
all_data = []

for symbol, company in stocks.items():
#Menggunakan Ticker().history() agar data tidak MultiIndex
    df = yf.Ticker(symbol).history(
        start="2020-01-01",
        end="2025-12-31"
    )

    df = df.reset_index()

    df["symbol"] = symbol
    df["company_name"] = company

    df = df.rename(columns={
        "Date": "price_date",
        "Open": "open_price",
        "High": "high_price",
        "Low": "low_price",
        "Close": "close_price",
        "Volume": "volume"
    })

    df = df[
        ["symbol", "company_name",
         "open_price", "high_price",
         "low_price", "close_price",
         "volume", "price_date"]
    ]

    all_data.append(df)

final_data = pd.concat(all_data, ignore_index=True)
```

#Eksport ke CSV

```python
final_data.to_csv(
    'lq45_stock_data.csv',
    index=False
)

print("berhasil")
```
